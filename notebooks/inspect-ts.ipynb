# Inspect the alexchilton two-stage members

Print the real folder layout and the head tensor names so the loader is written against what is actually there.

```python
"""Print the exact layout and head keys of the alexchilton two-stage members, so the loader is written
against what is actually there rather than what I assume."""
import glob, json, os, sys
import torch
from safetensors import safe_open

for name in ("ntl-ts-q3-8b", "ntl-ts-gemma2-9b"):
    hits = glob.glob(f"/kaggle/input/**/{name}/meta.json", recursive=True)
    print(f"===== {name}: {hits[:1] or 'NOT FOUND'}")
    if not hits:
        continue
    root = os.path.dirname(hits[0])
    print("  meta.json:", json.load(open(hits[0])))
    for p in sorted(glob.glob(f"{root}/**/*", recursive=True)):
        if os.path.isfile(p):
            print(f"    {os.path.relpath(p, root):<44} {os.path.getsize(p):>14,}")
    for hp in glob.glob(f"{root}/**/heads.safetensors", recursive=True):
        with safe_open(hp, framework="pt") as f:
            ks = list(f.keys())
        print(f"  heads keys in {os.path.basename(hp)} ({len(ks)}): {ks[:12]}")
```
